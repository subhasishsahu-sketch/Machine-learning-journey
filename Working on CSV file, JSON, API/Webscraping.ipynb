```python
import pandas as pd
import requests

# used for web scraping
from bs4 import BeautifulSoup
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}
x=requests.get("https://www.ambitionbox.com/list-of-companies?page=1",headers=headers)
```

```python
soup=BeautifulSoup(x.text,"lxml")
soup.find_all("h2")
```

```python
#cleaning
name=""
data_c=[]
for i in soup.find_all("h2",class_="companyCardWrapper__companyName"):
    name = i.get_text(strip=True)
    data_c.append(name)

data_c
```

```python
# deriving the ratting from p tag
data_r=[]
# print(soup.find_all("div"))
for i in soup.find_all("p",class_="subNavDescription"):
    name = i.get_text(strip=True)
    data_r.append(name)

data_r
```

```python
ratting=[]
for i in soup.find_all("div",class_="companyCardWrapper__companyRating"):
    name = i.get_text(strip=True)
    ratting.append(name)

ratting
```

```python
rattingcount=[]
for i in soup.find_all("span",class_="companyCardWrapper__companyRatingCount"):
    name = i.get_text(strip=True)
    rattingcount.append(name)

rattingcount
```
