# Handeling date and time

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Database.csv",)
df=df[["Date","Time"]]
```

```python
df.info()
```

```python
# converting the string date column to date time column
df['Date'] = pd.to_datetime(df['Date'], utc=True, errors='coerce')
```

```python
df.columns
```

```python
df
```

## Extract Year

```python
df['date_year'] = df['Date'].dt.year
df
```

## Extract Month

```python
df['date_month'] = df['Date'].dt.month
df
```

```python
#With names of the months

df['date_month_name'] = df['Date'].dt.month_name()
df
```

## Extract Days

```python
df['date_days'] = df['Date'].dt.day
df
```

## Extract Day of week

```python
df['date_DayOfWeek'] = df['Date'].dt.dayofweek
df
```

## Extract Day of week -name

```python
df['date_DayOfWeek_name'] = df['Date'].dt.day_name()
df
```

## Is it the weekend

```python
df['date_is_weekend'] = np.where(df['date_DayOfWeek_name'].isin(['Sunday', 'Saturday']), 1, 0)
df
```

## Extract week of the year

```python
df['date_week'] = df['Date'].dt.isocalendar().week
df
```

## Extract Quater

```python
df['date_quarter'] = df['Date'].dt.quarter
df
```

## Extract tome elapsed between dates

```python
import datetime
today = datetime.datetime.today()
today # Todays data stamp
```

```python
today = pd.Timestamp.now(tz='UTC')
today-df["Date"]
```

```python
(today-df["Date"]).dt.days
```

```python
# years passed
df['years_diff'] =(today - df['Date']).dt.days / 365
df
```

```python
#months passed
df['months_diff'] =(today - df['Date']).dt.days / 30
df
```

## For extract of time

```python
df['Time'] = pd.to_datetime(df['Time'], errors='coerce')

df['hour'] = df['Time'].dt.hour
df['min'] = df['Time'].dt.minute
df['sec'] = df['Time'].dt.second

df
```
