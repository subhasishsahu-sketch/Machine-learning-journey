# Multivarient Analysis
Multivariate analysis studies how multiple variables interact and influence each other simultaneously.

```python
import pandas as pd
import seaborn as sns
```

## Dataset
different dataset... some are inbuilt and are imported from the sns (seaborn)

```python
tips=sns.load_dataset("tips")
```

```python
tips.head()
```

```python
tit=pd.read_csv("customer.csv")
tit
```

```python
titan=pd.read_csv("Titanic-Dataset.csv")
titan
```

```python
flights = sns.load_dataset("flights")
flights
```

```python
iris = sns.load_dataset("iris")
iris
```

## Scatterplot ( numerical - numerical)

```python
sns.scatterplot(x=tips["total_bill"],y=tips["tip"],hue=tips["sex"],style=tips["smoker"],size=tips["size"])
```

## Bar Plot (Numerical - Categorical)

```python
sns.barplot(x=tit["gender"],y=tit["age"])
```

```python
sns.barplot(y=tit["age"],x=tit["education"],hue=tit["gender"])
```

## Box plot

```python
sns.boxplot(x=tit["gender"],y=tit["age"])
```

## Distplot (Numerical - Categorical)

```python
sns.distplot(tit["purchased"]=="Yes",hist=False)
sns.distplot(tit["purchased"]=="No",hist=False)
```

```python
tit.head()
```

## HeatMap (Categorical-Categorical)

```python
sns.heatmap(pd.crosstab(tit["gender"],tit["review"]))
```

```python
titan
```

```python
titan.groupby("Pclass")["Age"].mean()
```

```python
titan.groupby("Pclass")["Age"].mean().plot(kind="bar")
```

```python
titan.groupby("Sex")["Age"].mean()
```

```python
titan.groupby("Pclass")["Age"].mean().plot(kind="bar")
```

## Clustermap (Categorical-Categorical)

```python
sns.clustermap(pd.crosstab(titan["Parch"],titan["Survived"]))
```

## Pairplot 

```python
sns.pairplot(iris)
```

## Line plot

It is only ploted or mostly plotted for the time vs (any thing in y axis)

```python
new = flights.groupby('year').sum()

sns.lineplot(x=new['year'], y=new['passengers'])
```
