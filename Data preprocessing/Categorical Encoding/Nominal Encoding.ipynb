```python
fd = df[["months_loan_duration", "credit_history","purpose","installment_rate","personal_status","property","age","amount"]]

from sklearn.model_selection import train_test_split


x_train,x_test,y_train,y_test=train_test_split (fd.iloc[:,:9],fd.iloc[:,-1],test_size=0.3)
```

```python
x_train,"----------",y_train
```

```python
from sklearn.preprocessing import OneHotEncoder

# inside the function the parameter is passed to remove the first column of the mosified data 
# changing the datatype from dr=efult double into int 32
ohe=OneHotEncoder(drop="first",sparse_output=False,dtype =np.int32) # object
```

```python
x_train_new = ohe.fit_transform(x_train[["credit_history","purpose","personal_status","property"]])

x_test_new = ohe.transform(x_test[["credit_history","purpose","personal_status","property"]])
```

```python
x_test_new,"---------------",x_train_new
```

```python
# Adding the processed data into the data set 

x_train_final = np.hstack((
    x_train[["months_loan_duration","installment_rate","age","amount"]].values,
    x_train_new
))

# colomn reduced as the passing of drop parameter in the ohe object 
print (x_train_final.shape)
x_train_final
```

```python
# Adding the processed data into the data set of train

x_test_final = np.hstack((
    x_test[["months_loan_duration","installment_rate","age","amount"]].values,
    x_test_new
))

# colomn reduced as the passing of drop parameter in the ohe object 
print(x_test_final.shape)
x_test_final
```
