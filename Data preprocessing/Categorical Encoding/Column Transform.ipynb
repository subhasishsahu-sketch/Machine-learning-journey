# Column Transform

## Import of lib

```python
from sklearn.compose import ColumnTransformer
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

## Column transform use 

```python
df = pd.read_csv("customer.csv")
df
```

```python
# creating the object 

# In the transformer parameter the columns are passed in which pressprocessing are to be done 
# in reminders parameter there is 2 option as passthrough/drop
transformer = ColumnTransformer(transformers=[
    ("tnfs1",SimpleImputer(),["age"]), #although the simple imputer is not required as it has no null values but is used to aee as example for the future impliment
     ("tnfs2",OrdinalEncoder( categories=[["Poor","Average","Good"],["School","UG","PG"]]),["review","education"]),
     ("tnfs3",OneHotEncoder(sparse_output=False,drop="first"),["gender","purchased"])
],remainder="passthrough")
```

```python
# applying on the real data
x=transformer.fit_transform(df)
x
```
