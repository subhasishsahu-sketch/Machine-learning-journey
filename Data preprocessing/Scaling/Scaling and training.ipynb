```python
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("student_data.csv")
```

```python
df=df[["age","absences","Medu"]]
df
```

```python
# the parmaeter are given line (x-given data,y- target,testsize (double like 0.3- means 30 % of data will be tested),random_state=(integer like 0- for no remdomness and 42 is for more randomness))

X_train, X_test, y_train, y_test = train_test_split(df,
    df['absences'],
    test_size=0.3,
    random_state=0
)

X_train.shape, X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# fit the scaler to the train set, it will learn the parameters
scaler.fit(X_train)

# transform train and test sets
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

scaler.mean_
```
